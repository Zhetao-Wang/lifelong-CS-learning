```python
from github import Github

# 创建一个连接 GitHub 的客户端
with Github(timeout=15) as client:
    # 向 GitHub 查询 Python 官方源码仓库
    repo = client.get_repo("Zhetao-Wang/dataprof")

    print("项目：", repo.full_name)
    print("简介：", repo.description)
    print("Star 数：", repo.stargazers_count)
    print("项目地址：", repo.html_url)
```
